# GC Example 10: Relaxation Time and Dynamical State

**EPS Research — Milky Way GC Corpus v1.4.0**

Half-mass relaxation time t_rh indicates how 'relaxed' a cluster is.
Core-collapsed clusters have the shortest relaxation times.

**Corpus:** Flynn (2026), EPS Research Milky Way Globular Cluster Corpus v1.4.0  
**Sources:** Harris (1996/2010), Vasiliev & Baumgardt (2021), Baumgardt et al. (2023), Schiavon et al. (2024) APOGEE DR17  
**Dependencies:** Python 3, numpy, matplotlib

```python
# ── Colab setup: auto-download corrected GC corpus ─────────────
import os, sys
IN_COLAB = 'google.colab' in sys.modules

if IN_COLAB:
    import urllib.request
    GC_PATH = 'harris_gc_corpus_v1.4.0.jsonl'
    GC_URL = 'https://raw.githubusercontent.com/eps-research/rag-corpus-series/9734a42a025900bc26ec69373628114841e2b2e0/examples/gc/harris_gc_corpus_v1.4.0.jsonl'
    if not os.path.exists(GC_PATH):
        print(f"Downloading {GC_PATH}...")
        urllib.request.urlretrieve(GC_URL, GC_PATH)
        print(f"  ✓ {GC_PATH}")
    else:
        print(f"  Already present: {GC_PATH}")
    print("Ready.")
else:
    GC_PATH = 'harris_gc_corpus_v1.4.0.jsonl'
    print(f"Running locally — GC corpus: {GC_PATH}")
```

```python
import json
import numpy as np
import matplotlib.pyplot as plt

clusters = []
with open(GC_PATH) as f:
    for line in f:
        clusters.append(json.loads(line))
print(f"Total clusters: {len(clusters)}")
```

```python
data=[]
for c in clusters:
    b=c.get('baumgardt2023',{})
    h=c.get('structure',{})
    if b and b.get('log_trh_yr') and h:
        data.append({'cluster':c['cluster_id'],
                     'log_trh':b['log_trh_yr'],
                     'core_collapsed':h.get('core_collapsed',False),
                     'mass':b.get('mass_msun',0)})
cc=[d for d in data if d['core_collapsed']]
nc=[d for d in data if not d['core_collapsed']]
print(f"Total with t_rh: {len(data)}")
print(f"Core-collapsed: {len(cc)}")
print(f"Not core-collapsed: {len(nc)}")
print(f"Mean log t_rh (CC): {np.mean([d['log_trh'] for d in cc]):.2f} yr")
print(f"Mean log t_rh (non-CC): {np.mean([d['log_trh'] for d in nc]):.2f} yr")
fig,ax=plt.subplots(figsize=(7,4))
ax.hist([d['log_trh'] for d in nc],bins=15,alpha=0.7,color='#2166ac',
        label=f'Normal ({len(nc)})',edgecolor='white')
ax.hist([d['log_trh'] for d in cc],bins=10,alpha=0.8,color='#d62728',
        label=f'Core-collapsed ({len(cc)})',edgecolor='white')
ax.set_xlabel(r'log$_{10}$ $t_{\rm rh}$ (yr)',fontsize=11); ax.set_ylabel('N',fontsize=11)
ax.set_title('Half-Mass Relaxation Time\nEPS Research GC Corpus v1.4.0',fontsize=10)
ax.legend(); plt.tight_layout(); plt.savefig('gc10_relaxation.png',dpi=150,bbox_inches='tight'); plt.show()
```
