# GC Example 20: Mass Segregation Parameters

**EPS Research — Milky Way GC Corpus v1.4.0**

Baumgardt provides eta_c and eta_h — mass segregation parameters
from Trenti & van der Marel (2013).
Negative eta means more massive stars concentrated toward center.

**Corpus:** Flynn (2026), EPS Research Milky Way Globular Cluster Corpus v1.4.0  
**Sources:** Harris (1996/2010), Vasiliev & Baumgardt (2021), Baumgardt et al. (2023), Schiavon et al. (2024) APOGEE DR17  
**Dependencies:** Python 3, numpy, matplotlib

```python
# ── Colab setup: auto-download corrected GC corpus ─────────────
import os, sys
IN_COLAB = 'google.colab' in sys.modules

if IN_COLAB:
    import urllib.request
    GC_PATH = 'harris_gc_corpus_v1.4.0.jsonl'
    GC_URL = 'https://raw.githubusercontent.com/eps-research/rag-corpus-series/9734a42a025900bc26ec69373628114841e2b2e0/examples/gc/harris_gc_corpus_v1.4.0.jsonl'
    if not os.path.exists(GC_PATH):
        print(f"Downloading {GC_PATH}...")
        urllib.request.urlretrieve(GC_URL, GC_PATH)
        print(f"  ✓ {GC_PATH}")
    else:
        print(f"  Already present: {GC_PATH}")
    print("Ready.")
else:
    GC_PATH = 'harris_gc_corpus_v1.4.0.jsonl'
    print(f"Running locally — GC corpus: {GC_PATH}")
```

```python
import json
import numpy as np
import matplotlib.pyplot as plt

clusters = []
with open(GC_PATH) as f:
    for line in f:
        clusters.append(json.loads(line))
print(f"Total clusters: {len(clusters)}")
```

```python
data=[(c['cluster_id'],c['baumgardt2023']['eta_c'],c['baumgardt2023']['eta_h'],
        c['baumgardt2023']['mass_msun'])
      for c in clusters if c.get('baumgardt2023') and
      c['baumgardt2023'].get('eta_c') is not None and
      c['baumgardt2023'].get('eta_h') is not None]
eta_c=[d[1] for d in data]; eta_h=[d[2] for d in data]; mass=[d[3] for d in data]
print(f"Clusters with mass segregation data: {len(data)}")
print(f"eta_c range: {min(eta_c):.2f} -- {max(eta_c):.2f}")
print(f"eta_h range: {min(eta_h):.2f} -- {max(eta_h):.2f}")
fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].scatter(eta_c,eta_h,s=15,alpha=0.6,color='#9467bd')
axes[0].axhline(0,color='black',ls='--',lw=0.8,alpha=0.5)
axes[0].axvline(0,color='black',ls='--',lw=0.8,alpha=0.5)
axes[0].set_xlabel(r'$\eta_c$ (core)',fontsize=11); axes[0].set_ylabel(r'$\eta_h$ (half-mass)',fontsize=11)
axes[0].set_title('Mass Segregation Parameters',fontsize=10)
axes[1].scatter(np.log10(mass),eta_c,s=15,alpha=0.6,color='#9467bd')
axes[1].set_xlabel(r'log Mass ($M_\odot$)',fontsize=11); axes[1].set_ylabel(r'$\eta_c$',fontsize=11)
axes[1].set_title('Segregation vs Mass',fontsize=10)
plt.suptitle('Mass Segregation — EPS Research GC Corpus v1.4.0',fontsize=11)
plt.tight_layout(); plt.savefig('gc20_segregation.png',dpi=150,bbox_inches='tight'); plt.show()
```
