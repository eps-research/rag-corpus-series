# IntZ Example 19: Omega Withdrawal Audit
**EPS Research IntZ Kinematic Corpus v1.0**

The August 2026 correction withdrew the 166 KROSS omega values after the
Eq. 6 defect was identified. Current omega values are therefore null.

This notebook verifies that remediation state. Legacy omega values are
retained only for provenance and are not used for current scientific claims.


```python
# ── Colab setup: auto-download corpus from Zenodo ─────────────
import os, sys
IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    import urllib.request
    CORPORA = {
        'intz_corpus_v1b.json': 'https://zenodo.org/records/21841382/files/intz_corpus_v1b.json',
        'intz_corpus_v1b_flat.csv': 'https://zenodo.org/records/21841382/files/intz_corpus_v1b_flat.csv',
    }
    for filename, url in CORPORA.items():
        if not os.path.exists(filename):
            print(f"Downloading {filename}...")
            urllib.request.urlretrieve(url, filename)
            print(f"  ✓ {filename}")
        else:
            print(f"  Already present: {filename}")
    print("Ready.")
else:
    print("Running locally — corpus files loaded from working directory.")
```

```python
import json
import numpy as np
import matplotlib.pyplot as plt

# Load IntZ corpus
with open('intz_corpus_v1b.json') as f:
    data = json.load(f)
galaxies = data['galaxies']
print(f"Loaded {len(galaxies)} galaxies")
```

```python
from collections import Counter
import matplotlib.pyplot as plt

current = []
legacy = []
statuses = Counter()

for g in galaxies:
    om = g.get('omega', {})

    if om.get('omega_available') and om.get('omega_value_rad_gyr') is not None:
        current.append(om['omega_value_rad_gyr'])

    if om.get('legacy_omega_value') is not None:
        legacy.append(om['legacy_omega_value'])

    if om.get('legacy_omega_status'):
        statuses[om['legacy_omega_status']] += 1

print(f"Current usable omega values: {len(current)}")
print(f"Preserved legacy omega values: {len(legacy)}")
print("Legacy status counts:", dict(statuses))

assert len(current) == 0
assert len(legacy) == 166

fig, ax = plt.subplots(figsize=(7,4))
ax.bar(['Current usable omega', 'Withdrawn legacy omega'],
       [len(current), len(legacy)])
ax.set_ylabel('N galaxies')
ax.set_title('IntZ Omega Remediation Audit')
plt.tight_layout()
plt.savefig('intz_nb19_omega_withdrawal_audit.png',
            dpi=150, bbox_inches='tight')
plt.show()

print()
print("AUDIT PASS:")
print("  Current IntZ omega values are null.")
print("  166 withdrawn KROSS legacy values remain for provenance only.")
print("  No omega-outlier inference is performed.")
```
