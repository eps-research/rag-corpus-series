# IntZ Example 12: Stellar-Mass–Rotation Relation at z ~ 0.9

**EPS Research IntZ Kinematic Corpus v1.0** | Flynn, D.C. (2026)

Examines the observed relation between stellar mass and circular velocity
for KROSS Tier-1 galaxies at z ~ 0.6–1.0.

This is a **stellar-mass–rotation relation**, not a baryonic
Tully-Fisher relation: the IntZ quantity used here is stellar mass only.

> **Note (August 2026):** This notebook previously showed omega vs
> stellar mass. Omega values have been withdrawn — see
> [CORRECTIONS.md](../../CORRECTIONS.md). The obsolete approximate
> SPARC comparison has also been removed.


```python
import json
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

with open('intz_corpus_v1b.json') as f:
    data = json.load(f)

rows = []

for g in data['galaxies']:
    if g.get('identifiers', {}).get('survey') != 'KROSS':
        continue
    if g.get('metadata', {}).get('quality_tier') != 1:
        continue

    z = g.get('redshift', {}).get('z_spec')
    vc = g.get('kinematics', {}).get('Vc_kms')
    log_ms = g.get('stellar_properties', {}).get('log_mass_msun')

    if z is None or vc is None or log_ms is None:
        continue

    rows.append({
        'z': float(z),
        'Vc': float(vc),
        'log_ms': float(log_ms),
    })

print(f"KROSS Tier-1 with Vc + stellar mass: {len(rows)} galaxies")
```

```python
log_ms = np.array([r['log_ms'] for r in rows])
vc = np.array([r['Vc'] for r in rows])
z = np.array([r['z'] for r in rows])

assert len(rows) > 0

fig, ax = plt.subplots(figsize=(8, 6))

sc = ax.scatter(
    log_ms,
    vc,
    c=z,
    cmap='plasma',
    s=20,
    alpha=0.65
)

plt.colorbar(sc, ax=ax, label='Redshift z')

ax.set_xlabel(r'log $M_*$ ($M_\odot$)', fontsize=12)
ax.set_ylabel('Vc (km/s)', fontsize=12)
ax.set_title(
    'KROSS Stellar-Mass–Rotation Relation at z ~ 0.9',
    fontsize=11
)

plt.tight_layout()
plt.savefig(
    'intz_nb12_stellar_mass_rotation_z09.png',
    dpi=150,
    bbox_inches='tight'
)
plt.show()

print(f'N galaxies with Vc + stellar mass: {len(rows)}')
print(f'Stellar mass range: {log_ms.min():.2f} to {log_ms.max():.2f} log Msun')
print(f'Vc range: {vc.min():.1f} to {vc.max():.1f} km/s')
print(f'Median stellar mass: {np.median(log_ms):.2f} log Msun')
print(f'Median Vc: {np.median(vc):.1f} km/s')
```
