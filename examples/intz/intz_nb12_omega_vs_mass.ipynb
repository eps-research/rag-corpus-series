# IntZ Example 12: Tully-Fisher Relation at z ~ 0.9

**EPS Research IntZ Kinematic Corpus v1.0** | Flynn, D.C. (2026)

The baryonic Tully-Fisher relation (circular velocity vs stellar mass)
for KROSS Tier-1 galaxies at z ~ 0.6–1.0, compared to the z=0 SPARC
reference.

> **Note (August 2026):** This notebook previously showed omega vs
> stellar mass. Omega values have been withdrawn — see
> [CORRECTIONS.md](../../CORRECTIONS.md). Replaced with Tully-Fisher relation.


```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import csv, numpy as np

# Load IntZ corpus flat CSV
rows = []
with open('intz_corpus_v1b_flat.csv') as f:
    for row in csv.DictReader(f):
        def fv(k):
            try: return float(row[k])
            except: return None
        rows.append({
            'z':      fv('z_spec'),
            'Vc':     fv('Vc_kms'),
            'sigma':  fv('sigma0_kms'),
            'vos':    fv('v_over_sigma'),
            'log_ms': fv('log_mstar'),
            'Reff':   fv('Reff_kpc'),
            'survey': row['survey'],
            'tier':   row['quality_tier'],
        })

kross1 = [r for r in rows if r['survey']=='KROSS' and r['tier']=='1' and r['Vc']]
print(f'KROSS Tier-1 with Vc: {len(kross1)} galaxies')
```

```python
tf = [r for r in kross1 if r['log_ms']]
log_ms = [r['log_ms'] for r in tf]
vc     = [r['Vc']     for r in tf]

# SPARC z=0 reference (approximate TFR slope from McGaugh 2012)
ms_range = np.linspace(9.0, 11.5, 50)
vc_sparc = 10**((ms_range - 1.5) / 3.5) * 50  # approximate

fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(ms_range, vc_sparc, 'g--', lw=2, label='SPARC z=0 TFR (approximate)')
sc = ax.scatter(log_ms, vc, c=[r['z'] for r in tf],
                cmap='plasma', s=20, alpha=0.6, vmin=0.6, vmax=1.05,
                label=f'KROSS z~0.9 (N={len(tf)})')
plt.colorbar(sc, ax=ax, label='Redshift z')
ax.set_xlabel(r'log $M_*$ ($M_\odot$)', fontsize=12)
ax.set_ylabel('Vc (km/s)', fontsize=12)
ax.set_title('Tully-Fisher Relation at z ~ 0.9\n'
             'KROSS Tier-1 vs SPARC z=0 reference', fontsize=11)
ax.legend(fontsize=9)
plt.tight_layout()
plt.savefig('intz_nb12_tully_fisher_z09.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'N galaxies with Vc + Mstar: {len(tf)}')
print(f'Stellar mass range: {min(log_ms):.2f} to {max(log_ms):.2f} log Msun')
print(f'Vc range: {min(vc):.1f} to {max(vc):.1f} km/s')
```
