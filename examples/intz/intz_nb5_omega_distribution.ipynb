# IntZ Example 5: Circular Velocity Distribution

**EPS Research IntZ Kinematic Corpus v1.0** | Flynn, D.C. (2026)

Distribution of circular velocities for 166 KROSS Tier-1 galaxies
at z ~ 0.6–1.0 (beam-smear corrected).

> **Note (August 2026):** This notebook previously showed the omega
> distribution. Omega values have been withdrawn — see
> [CORRECTIONS.md](../../CORRECTIONS.md). Replaced with Vc distribution.


```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import csv, numpy as np

# Load IntZ corpus flat CSV
rows = []
with open('intz_corpus_v1b_flat.csv') as f:
    for row in csv.DictReader(f):
        def fv(k):
            try: return float(row[k])
            except: return None
        rows.append({
            'z':      fv('z_spec'),
            'Vc':     fv('Vc_kms'),
            'sigma':  fv('sigma0_kms'),
            'vos':    fv('v_over_sigma'),
            'log_ms': fv('log_mstar'),
            'Reff':   fv('Reff_kpc'),
            'survey': row['survey'],
            'tier':   row['quality_tier'],
        })

kross1 = [r for r in rows if r['survey']=='KROSS' and r['tier']=='1' and r['Vc']]
print(f'KROSS Tier-1 with Vc: {len(kross1)} galaxies')
```

```python
vc = [r['Vc'] for r in kross1]
z  = [r['z']  for r in kross1]

fig, axes = plt.subplots(1, 2, figsize=(11, 5))

# Panel 1: Vc histogram
ax = axes[0]
ax.hist(vc, bins=25, color='#ff7f0e', alpha=0.8, edgecolor='white')
ax.axvline(np.median(vc), color='red', lw=2, ls='--',
           label=f'Median = {np.median(vc):.1f} km/s')
ax.axvline(np.mean(vc), color='black', lw=1.5, ls=':',
           label=f'Mean = {np.mean(vc):.1f} km/s')
ax.set_xlabel('Circular Velocity Vc (km/s)', fontsize=12)
ax.set_ylabel('N galaxies', fontsize=12)
ax.set_title('KROSS Tier-1 Vc Distribution\nz ~ 0.6–1.0 (N=166)', fontsize=11)
ax.legend(fontsize=10)

# Panel 2: Vc vs redshift
ax2 = axes[1]
sc = ax2.scatter(z, vc, c=vc, cmap='plasma', s=20, alpha=0.7)
plt.colorbar(sc, ax=ax2, label='Vc (km/s)')
ax2.axhline(np.median(vc), color='red', lw=1.5, ls='--',
            label=f'Median {np.median(vc):.1f} km/s')
ax2.set_xlabel('Redshift z', fontsize=12)
ax2.set_ylabel('Vc (km/s)', fontsize=12)
ax2.set_title('Vc vs Redshift\nKROSS Tier-1', fontsize=11)
ax2.legend(fontsize=9)

plt.suptitle('IntZ Corpus v1.0 — Circular Velocity Distribution\n'
             'KROSS Tier-1, beam-smear corrected', fontsize=11)
plt.tight_layout()
plt.savefig('intz_nb5_vc_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Vc statistics:')
print(f'  N:      {len(vc)}')
print(f'  Median: {np.median(vc):.1f} km/s')
print(f'  Mean:   {np.mean(vc):.1f} km/s')
print(f'  Std:    {np.std(vc):.1f} km/s')
print(f'  Min:    {min(vc):.1f} km/s')
print(f'  Max:    {max(vc):.1f} km/s')
```
